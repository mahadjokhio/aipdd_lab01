# Real-Time Object Detection Using YOLO

## Project Overview

### Problem Statement
Manual monitoring of images and video is slow, tiring, and error prone.
This project builds an AI system that **detects and labels objects automatically** in images and video.

### Core Objectives
- Set up a clean and reproducible AI project environment
- Train and test an object detection model using **Ultralytics YOLO**
- Evaluate the model using metrics such as precision, recall, and mAP
- Document the full workflow so others can repeat it

### Environment Requirements
- **Python:** 3.12
- **Libraries:** numpy, opencv-python, matplotlib, ultralytics, jupyter
- **Environment name:** `aipdd_env`

### Setup Commands
```bash
python -m venv aipdd_env
aipdd_env\Scripts\activate.bat
pip install -r requirements.txt
```
